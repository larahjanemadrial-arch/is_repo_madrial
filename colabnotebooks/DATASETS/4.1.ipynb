```python
import pandas as pd
import numpy as np
```

```python
df_gdp = pd.read_excel('GDP(PPP)_LifeExpectancy.xlsx')

df_gdp.columns = df_gdp.columns.str.strip()

for column in ["GDP", "Life expectancy", "Population"]:
    if column in df_gdp.columns:
        df_gdp[column] = pd.to_numeric(df_gdp[column], errors="coerce")

# 1. What are the dimensions of the DataFrame?
print("1. Dimensions (rows, columns):", df_gdp.shape)

# 2. How many valid GDP values are there?
print("2. Valid GDP values:", df_gdp["GDP"].count())

# 3. What is the average life expectancy for all countries?
print("3. Average life expectancy:", df_gdp["Life expectancy"].mean())

# 4. Sort the DataFrame by Country and assign it to a new DataFrame
df_gdp_sorted = df_gdp.sort_values("Country").reset_index(drop=True)
print("\n4. DataFrame sorted by Country:")
display(df_gdp_sorted)

# 5. What are the top 5 countries with the highest GDP?
print("5. Top 5 countries with the highest GDP:")
display(df_gdp.nlargest(5, "GDP")[["Country", "GDP"]])

# 6. What are the top 5 countries with the smallest population?
print("6. Top 5 countries with the smallest population:")
display(df_gdp.nsmallest(5, "Population")[["Country", "Population"]])
```
