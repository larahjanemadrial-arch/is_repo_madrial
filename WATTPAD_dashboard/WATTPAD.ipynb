```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

def generate_wattpad_data():
    rows = []

    popular_authors = ["cecelib", "jonaxx", "inksteady", "inerryss", "magietearjerky"]
    other_authors = ["DreamerWriter", "ShadowPen", "LunaInk", "AuthorX", "StoryQueen"]

    genres = ["Romance", "Fantasy", "Teen Fiction", "Horror", "Mystery",
              "Science Fiction", "Action", "Historical Fiction", "Paranormal", "Fanfiction"]

    review_categories = ['Unsatisfied', 'Satisfied', 'Good', 'Very Good', 'Excellent']

    comments_pool = [
        "This story is so addictive!", "Made me cry in chapter 15", "Best story ever!",
        "The plot twist was insane!", "I love this couple so much ❤️", "Update faster please 😭",
        "This healed me", "One of the best on Wattpad", "The tension is perfect",
        "Absolutely recommend!", "My new comfort story"
    ]

    np.random.seed(42)
    random.seed(42)

    for i in range(2000):
        # Author
        if i < 300:
            author = popular_authors[i % len(popular_authors)]
        else:
            author = np.random.choice(popular_authors + other_authors)

        # Genre
        if author in ["cecelib", "jonaxx", "magietearjerky"]:
            genre = "Romance"
        elif author == "inerryss":
            genre = "Fantasy"
        elif author == "inksteady":
            genre = "Teen Fiction"
        else:
            genre = np.random.choice(genres)

        title = f"{author.capitalize()} Story {(i % 90) + 1}"
        code = f"WP-{str(i+1).zfill(5)}"
        reads = np.random.randint(12000, 1350000)

        review = np.random.choice(review_categories, p=[0.08, 0.18, 0.27, 0.32, 0.15])

        num_comments = random.randint(5, 25)
        selected_comments = random.sample(comments_pool, min(num_comments, len(comments_pool)))
        comments_text = " | ".join(selected_comments)

        rows.append({
            'Code': code,
            'Title': title,
            'Author': author,
            'Genre': genre,
            'Reads': reads,
            'Review': review,
            'Comments': comments_text,
            'Comment_Count': num_comments
        })

    return pd.DataFrame(rows)

df = generate_wattpad_data()
```

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
import plotly.express as px

# Reader Reviews Pie Chart
review_data = df['Review'].value_counts().reset_index()
review_data.columns = ['Review', 'Count']

fig_review = px.pie(
    review_data,
    names='Review',
    values='Count',
    hole=0.45,
    title=" Reader Reviews",
    color_discrete_sequence=px.colors.diverging.RdYlGn
)

fig_review.update_traces(textinfo='percent+label')
fig_review.update_layout(height=500)
fig_review.show()
```

```python
genre_data = df.groupby('Genre')['Reads'].sum().reset_index()

fig_bar = px.bar(
    genre_data.sort_values('Reads', ascending=False),
    x='Genre',
    y='Reads',
    title="Reads by Genre",
    color='Reads',
    color_continuous_scale=px.colors.sequential.Oranges,
    text='Reads'
)

fig_bar.update_traces(
    texttemplate='%{text:,}',
    textposition='outside'
)

fig_bar.update_layout(
    height=550,
    xaxis_title="Genre",
    yaxis_title="Total Reads",
    title_font_size=22,
    title_x=0.5
)

fig_bar.show()
```

```python
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

app.index_string = '''
<!DOCTYPE html>
<html>
<head>
    {%metas%}
    <title>Wattpad Dashboard</title>
    {%favicon%}
    {%css%}
    <style>
        body { background-color: #fffaf0; font-family: 'Segoe UI', sans-serif; }
        h1, h4, h5 { color: #FF6600; font-weight: 700; }
        .form-control { border: 2px solid #FF8533; border-radius: 12px; font-size: 17px; }
        .form-control:focus { border-color: #FF6600; box-shadow: 0 0 0 0.25rem rgba(255, 102, 0, 0.25); }
        .card { border-radius: 16px; box-shadow: 0 6px 20px rgba(0,0,0,0.08); border: none; }
        .logo-w {
            font-size: 92px; font-weight: 900; color: #FF6600;
            font-family: 'Arial Black', sans-serif; line-height: 1; letter-spacing: -10px;
        }
        .header-title { font-size: 46px; font-weight: 700; color: #FF6600; }
    </style>
</head>
<body>
    {%app_entry%}
    <footer>{%config%}{%scripts%}{%renderer%}</footer>
</body>
</html>
'''

app.layout = dbc.Container([
    # HEADER
    dbc.Row([
        dbc.Col([
            html.Div([
                html.Span("W", className="logo-w me-4"),
                html.H1("Wattpad Stories Dashboard", className="header-title d-inline align-middle")
            ], className="d-flex align-items-center justify-content-center")
        ], width=12)
    ], className="mb-5 mt-4"),

    # SEARCH
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H4("Search by Author", className="mb-3"),
                    html.P("Try: cecelib, jonaxx, inksteady, inerryss, magietearjerky",
                           className="text-muted mb-3"),
                    dcc.Input(id="author-input", type="text", placeholder="Type author name here...",
                              debounce=True, className="form-control", style={"height": "60px"})
                ])
            ], className="mb-5")
        ], width=12)
    ]),

    # PIE CHARTS
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5(" Reads Distribution by Genre", className="mb-4"),
                    dcc.Graph(id="genre-pie", style={"height": "430px"})
                ])
            ])
        ], width=6),

        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5(" Reader Reviews Distribution", className="mb-4"),
                    dcc.Graph(id="review-pie", style={"height": "430px"})
                ])
            ])
        ], width=6)
    ], className="mb-5"),

    # TOP 5 STORIES
    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Top 5 Highest Read Stories", className="mb-4"),
                    dash_table.DataTable(
                        id="top5-table",
                        style_table={'overflowX': 'auto'},
                        style_header={'backgroundColor': '#FF6600', 'color': 'white', 'fontWeight': 'bold'},
                        style_cell={'padding': '12px', 'textAlign': 'left'}
                    )
                ])
            ])
        ], width=12)
    ], className="mb-4"),


    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardBody([
                    html.H5("Filtered Stories", className="mb-4"),
                    dash_table.DataTable(
                        id="search-table",
                        page_size=12,
                        style_table={'overflowX': 'auto'},
                        style_header={'backgroundColor': '#FF6600', 'color': 'white', 'fontWeight': 'bold'},
                        style_cell={'padding': '12px', 'textAlign': 'left', 'whiteSpace': 'normal'}
                    )
                ])
            ])
        ], width=12)
    ]),

], fluid=True, style={"maxWidth": "1500px"})

@app.callback([Output("genre-pie", "figure"), Output("review-pie", "figure")], Input("author-input", "value"))
def update_charts(author_filter):
    filtered = df.copy()
    if author_filter and author_filter.strip():
        filtered = filtered[filtered['Author'].str.contains(author_filter.strip(), case=False)]

    genre_fig = px.pie(filtered.groupby('Genre')['Reads'].sum().reset_index(),
                       names='Genre', values='Reads', hole=0.45,
                       title="Reads by Genre", color_discrete_sequence=px.colors.sequential.Oranges_r)
    genre_fig.update_traces(textinfo='percent+label')

    review_fig = px.pie(filtered['Review'].value_counts().reset_index(),
                        names='Review', values='count', hole=0.45,
                        title="Reader Reviews", color_discrete_sequence=px.colors.sequential.RdYlGn_r)
    review_fig.update_traces(textinfo='percent+label')

    return genre_fig, review_fig


@app.callback([Output("top5-table", "data"), Output("top5-table", "columns")], Input("author-input", "value"))
def update_top5(author_filter):
    filtered = df.copy()
    if author_filter and author_filter.strip():
        filtered = filtered[filtered['Author'].str.contains(author_filter.strip(), case=False)]

    top5 = filtered.nlargest(5, 'Reads')
    columns = [
        {"name": "Code", "id": "Code"},
        {"name": "Title", "id": "Title"},
        {"name": "Author", "id": "Author"},
        {"name": "Genre", "id": "Genre"},
        {"name": "Reads", "id": "Reads", "type": "numeric", "format": {"specifier": ","}},
        {"name": "Review", "id": "Review"}
    ]
    return top5.to_dict('records'), columns


@app.callback([Output("search-table", "data"), Output("search-table", "columns")], Input("author-input", "value"))
def update_search(author_filter):
    filtered = df.copy()
    if author_filter and author_filter.strip():
        filtered = filtered[filtered['Author'].str.contains(author_filter.strip(), case=False)]

    columns = [
        {"name": "Code", "id": "Code"},
        {"name": "Title", "id": "Title"},
        {"name": "Author", "id": "Author"},
        {"name": "Genre", "id": "Genre"},
        {"name": "Reads", "id": "Reads", "type": "numeric", "format": {"specifier": ","}},
        {"name": "Review", "id": "Review"},
        {"name": "Comments", "id": "Comments"}
    ]
    return filtered.to_dict('records'), columns


if __name__ == '__main__':
    app.run(jupyter_mode='inline', height=1300)
```
