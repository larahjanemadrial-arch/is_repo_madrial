```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://api.blockchain.info/charts/market-price?format=csv"

df_bitcoin = pd.read_csv(
    url,
    header=None,
    names=["Date", "Price"],
    parse_dates=["Date"]
)

df_bitcoin.to_csv("market-price.csv", index=False)
print("Saved a copy as market-price.csv")
```

```python
print("Rows and columns:", df_bitcoin.shape)
df_bitcoin.info()
display(df_bitcoin.head())
display(df_bitcoin.describe())
```

```python
df_bitcoin = df_bitcoin.sort_values("Date")

plt.figure(figsize=(12, 5))
plt.plot(df_bitcoin["Date"], df_bitcoin["Price"])
plt.title("Bitcoin Market Price")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.grid(True)
plt.tight_layout()
plt.show()
```
